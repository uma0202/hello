# Problem

A college student in an AI Bachelor's program wants to manage a small dataset of AI courses. The program should store course information, identify unique AI topics, maintain a queue of students waiting for a lab session, and display the collected information.

This example demonstrates the major Python data structures you asked for:

- **Variables** — store individual values.
- **String** — course names and student names.
- **List** — collection of AI courses.
- **Tuple** — fixed information about a course.
- **Dictionary** — course information using key-value pairs.
- **Set** — unique AI topics.
- **Queue** — students waiting for an AI lab.

```python
from collections import deque
```

```python
from collections import deque

# Variables
student_name = "Ahmed"
semester = 4
cgpa = 3.45

# String
course_name = "Artificial Intelligence"

# List
courses = [
    "Python Programming",
    "Artificial Intelligence",
    "Machine Learning",
    "Data Structures"
]

# Tuple
course_details = ("Artificial Intelligence", 4, "AI-204")

# Dictionary
student = {
    "name": student_name,
    "semester": semester,
    "cgpa": cgpa,
    "major": "Artificial Intelligence"
}

# Set
ai_topics = {
    "Machine Learning",
    "Neural Networks",
    "Computer Vision",
    "Machine Learning",
    "NLP"
}

# Queue
lab_queue = deque()

lab_queue.append("Ahmed")
lab_queue.append("Sara")
lab_queue.append("Omar")
lab_queue.append("Fatima")
```



# Display information
